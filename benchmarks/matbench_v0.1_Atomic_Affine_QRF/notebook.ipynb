# Atomic-Affine-Bagged-QRF

Composition-only Matbench steels. Fixed seed 417; no best-repeat selection. This notebook starts from raw chemistry and official folds and rebuilds 400 inner forests plus every active final forest. See info.json for exact package pins and retrospective-development limits. The actual notebook passed in a new Jupyter kernel with an empty fit directory; complete official fold dictionaries reproduce exactly.

The primary fits 128 Bayesian-bootstrap weighted MAE combinations of 32
ExtraTrees leaf-distribution conditional mean/median predictors. Each fit learns
nonnegative coefficients with sum at most two and a free intercept, using only
five-fold OOF predictions within the official outer training fold. Predictions
average all coefficient vectors and intercepts. The offset-only ablation keeps
coefficient sum one; the preceding method also fixed the intercept at zero.
The 16 forest configurations, all policies, and seeds were fixed before these
new outcomes, after earlier public-fold results were known. This is retrospective
development, not an untouched holdout. The fixed seed417 scores 75.5425582875 MPa;
repeat seeds0–4 average 75.7659913060 ± 0.4028456697 MPa (sample SD).

Established methods: [ExtraTrees](https://doi.org/10.1007/s10994-006-6226-1),
[Quantile Regression Forests](https://jmlr.org/papers/v7/meinshausen06a.html),
[Bayesian bootstrap](https://doi.org/10.1214/aos/1176345338), and
[Matbench](https://doi.org/10.1038/s41524-020-00406-3).
The result concerns this frozen implementation, not invention of these methods.
The bootstrap averages predictions; it does not establish posterior coverage.

Run under Python3.13 from this directory with the exact requirements in info.json.
The code cells below are unchanged from the completed fresh Jupyter replay.
The fresh_seed417 directory must be absent; three one-thread processes are used.
No fitted forests or OOF caches are supplied. The final assertion compares every
complete official fold dictionary, including coefficients, predictions and scores.


Environment check: install requirements.txt under CPython 3.13.16. The first code cell runs an environment/input/cache preflight before the unchanged fitting cells. The median is the frozen tolerance-defined lower median (CDF>=0.5-1e-12). No prediction tolerance is introduced.
